# Accident Forecasting with Machine Learning

This notebook demonstrates a basic machine learning approach to forecast the number of industrial accidents using historical data. We use a simple regression model to predict accident counts by month.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

```
# Load preprocessed dataset
url = 'https://raw.githubusercontent.com/huqingyuan314/DS5110-25Summer-Project/main/datasets/Industrial%20Safety%20and%20Health%20Analytics%20Database/IHMStefanini_industrial_safety_and_health_database_with_accidents_description.csv'
df = pd.read_csv(url)
df = df[df['Data'].str.match('^[0-9]{4}-', na=False)]
df['DateParsed'] = pd.to_datetime(df['Data'], errors='coerce')
df = df.dropna(subset=['DateParsed'])
df['YearMonth'] = df['DateParsed'].dt.to_period('M').astype(str)
monthly_counts = df.groupby('YearMonth').size().reset_index(name='AccidentCount')
```

## Model Training and Evaluation

```
# Prepare data
monthly_counts['MonthIndex'] = np.arange(len(monthly_counts))
X = monthly_counts[['MonthIndex']]
y = monthly_counts['AccidentCount']

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Linear Regression
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)

# Decision Tree
tree = DecisionTreeRegressor(max_depth=4)
tree.fit(X_train, y_train)
y_pred_tree = tree.predict(X_test)
```

```
# Evaluation Metrics
print("Linear Regression")
print("MSE:", mean_squared_error(y_test, y_pred_lr))
print("R²:", r2_score(y_test, y_pred_lr))

print("\nDecision Tree")
print("MSE:", mean_squared_error(y_test, y_pred_tree))
print("R²:", r2_score(y_test, y_pred_tree))
```

```
# Visualization
plt.figure(figsize=(12,5))
plt.plot(monthly_counts['YearMonth'], y, label='Actual')
plt.plot(monthly_counts['YearMonth'].iloc[-len(y_test):], y_pred_lr, label='Linear Regression')
plt.plot(monthly_counts['YearMonth'].iloc[-len(y_test):], y_pred_tree, label='Decision Tree', linestyle='dashed')
plt.xticks(rotation=45)
plt.title('Accident Count Forecast')
plt.legend()
plt.tight_layout()
plt.show()
```

## Conclusion
Both models offer a simple method to forecast monthly accident trends. While the Decision Tree can capture non-linear trends better, Linear Regression provides a clear baseline. Further improvements can be made using more features and real-time data.
